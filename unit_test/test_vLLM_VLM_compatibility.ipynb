```python
# Cell 2: Model Registry + Test
from vllm import LLM, SamplingParams
import gc
import torch

# All models from model_inference_vLLM.py
VLLM_MODEL_PATHS = {
    # Qwen (1-3)
    "Qwen25_VL_3B": "Qwen/Qwen2.5-VL-3B-Instruct",
    "Qwen25_VL_7B": "Qwen/Qwen2.5-VL-7B-Instruct",
    "Qwen2_VL_2B": "Qwen/Qwen2-VL-2B-Instruct",
    # LLaVA (4-5)
    "LLAVA_1pt5_7B": "llava-hf/llava-1.5-7b-hf",
    "LLAVA_v1pt6_Mistral_7B": "llava-hf/llava-v1.6-mistral-7b-hf",
    # InternVL (6-8)
    "InternVL3_1B": "OpenGVLab/InternVL3-1B",
    "InternVL3_2B": "OpenGVLab/InternVL3-2B",
    "InternVL25_4B": "OpenGVLab/InternVL2.5-4B",
    # PaliGemma (9)
    "PaliGemma_VL_3B": "google/paligemma-3b-mix-224",
    # New models (10-16)
    "MiniCPM_V_2pt5": "openbmb/MiniCPM-V-2_5",
    "DeepSeek_VL2_Small": "deepseek-ai/deepseek-vl2-small",
    "H2OVL_Mississippi_2B": "h2oai/h2ovl-mississippi-2b",
    "Phi35_Vision": "microsoft/Phi-3.5-vision-instruct",
    "Mono_InternVL_2B": "OpenGVLab/Mono-InternVL-2B",
    "MiniCPM_V_4": "openbmb/MiniCPM-V-4",
    "Molmo_7B": "allenai/Molmo-7B-D-0924",
}

# T4-16GB compatible models (≤4B params)
T4_COMPATIBLE = [
    "InternVL3_1B", "InternVL3_2B", "Qwen2_VL_2B",
    "H2OVL_Mississippi_2B", "Mono_InternVL_2B",
    "Qwen25_VL_3B", "PaliGemma_VL_3B", "MiniCPM_V_2pt5",
    "InternVL25_4B", "Phi35_Vision",
]

# T4-16GB: Skip 7B models (OOM risk)
T4_SKIP = ["Qwen25_VL_7B", "LLAVA_1pt5_7B", "LLAVA_v1pt6_Mistral_7B",
            "DeepSeek_VL2_Small", "MiniCPM_V_4", "Molmo_7B"]

def cleanup():
    gc.collect()
    torch.cuda.empty_cache()

def test_model(name, path):
    print(f"\n{'='*50}")
    print(f"Testing: {name}")
    print(f"Path: {path}")
    try:
        llm = LLM(
            model=path,
            trust_remote_code=True,
            gpu_memory_utilization=0.90,
            max_model_len=4096,
            dtype="half",
        )
        print(f"✅ {name} loaded successfully")
        del llm
        cleanup()
        return True
    except Exception as e:
        print(f"❌ {name} failed: {str(e)[:100]}")
        cleanup()
        return False

# Cell 3: Run tests
results = {}
for name in T4_COMPATIBLE:
    results[name] = test_model(name, VLLM_MODEL_PATHS[name])

# Summary
print(f"\n{'='*50}")
print("SUMMARY (T4-16GB)")
print("="*50)
for name, success in results.items():
    status = "✅" if success else "❌"
    print(f"{status} {name}")
```
